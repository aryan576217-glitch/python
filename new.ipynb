```python
import numpy as np
a=np.array([[10,20,30,40,50],[21,31,42,53,62],[43,54,76,93,96]])
b=np.array([[100,60,70,80,90],[53,62,83,94,71],[73,84,95,82,90]])
print(np.array(a+b))
print(np.sum(a))
print(np.min(a))
print(np.max(a))
print(np.mean(a))
print(a & b)
print(a[(a>20)&(a>40)])
```
